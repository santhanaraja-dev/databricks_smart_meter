```python
%sql
SELECT
    SUM(total_readings) AS total_readings
FROM smartmeter.gold.data_quality;
```

```python
%sql
SELECT
    ROUND(
        SUM(good_readings) * 100.0 / SUM(total_readings),
        2
    ) AS good_readings_pct
FROM smartmeter.gold.data_quality;
```

```python
%sql
SELECT
    SUM(bad_readings) AS bad_readings
FROM smartmeter.gold.data_quality;
```

```python
%sql
SELECT
    COALESCE(SUM(duplicate_readings), 0) AS duplicate_readings
FROM smartmeter.gold.data_quality;
```

```python
%sql
SELECT
    'Good' AS quality_status,
    SUM(good_readings) AS reading_count
FROM smartmeter.gold.data_quality

UNION ALL

SELECT
    'Bad' AS quality_status,
    SUM(bad_readings) AS reading_count
FROM smartmeter.gold.data_quality;
```

```python
%sql
SELECT
    area_name,
    SUM(total_readings) AS total_readings,
    SUM(good_readings) AS good_readings,
    SUM(bad_readings) AS bad_readings,
    ROUND(
        SUM(good_readings) * 100.0 / SUM(total_readings),
        2
    ) AS good_readings_pct
FROM smartmeter.gold.data_quality
GROUP BY area_name
ORDER BY good_readings_pct DESC;
```

```python
%sql
SELECT
    CASE
        WHEN last_reading_time IS NULL THEN 'No Reading'
        ELSE 'Reporting'
    END AS reading_status,
    COUNT(*) AS meter_count
FROM smartmeter.gold.meter_health
GROUP BY
    CASE
        WHEN last_reading_time IS NULL THEN 'No Reading'
        ELSE 'Reporting'
    END
ORDER BY meter_count DESC;
```

```python
%sql
SELECT
    meter_type,
    SUM(total_readings) AS total_readings,
    SUM(good_readings) AS good_readings,
    SUM(bad_readings) AS bad_readings,
    ROUND(
        SUM(good_readings) * 100.0 / SUM(total_readings),
        2
    ) AS good_readings_pct
FROM smartmeter.gold.data_quality
GROUP BY meter_type
ORDER BY good_readings_pct DESC;
```
