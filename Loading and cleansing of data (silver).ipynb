```python
%sql
CREATE VOLUME IF NOT EXISTS smartmeter.default.files;
```

# ALARMS TABLE

```python
alarms = spark.read.csv("/Volumes/smartmeter/default/files/alarms.csv",header=True,inferSchema=True)
display(alarms)

has_duplicates = alarms.count() > alarms.dropDuplicates().count()
print(has_duplicates)

alarms.printSchema()

from pyspark.sql.functions import col, sum
null_counts = alarms.select(
    [
        sum(col(c).isNull().cast("int")).alias(c)
        for c in alarms.columns
    ]
)

display(null_counts)
```

# METERS TABLE

```python
meters = spark.read.csv("/Volumes/smartmeter/default/files/meters.csv",header=True,inferSchema=True)
display(meters)

meters.printSchema()

has_duplicates = meters.count() > meters.dropDuplicates().count()
 
print(has_duplicates)

duplicate_meter_ids = (
    meters.groupBy("meter_id")
          .count()
          .filter("count > 1")
)

display(duplicate_meter_ids)

from pyspark.sql.functions import col, sum

null_counts = meters.select(
    [
        sum(col(c).isNull().cast("int")).alias(c)
        for c in meters.columns
    ]
)

display(null_counts)
```
