```python
%sql
SELECT
    COUNT(DISTINCT meter_id) AS reporting_electricity_meters
FROM smartmeter.gold.power_quality;
```

```python
%sql
SELECT
    COUNT(*) AS total_power_quality_readings
FROM smartmeter.gold.power_quality;
```

```python
%sql
SELECT
    timestamp,
    ROUND(AVG(value),2) AS avg_voltage
FROM smartmeter.gold.power_quality
WHERE parameter_name = 'Voltage (L1)'
GROUP BY timestamp
ORDER BY timestamp;
```

```python
%sql
SELECT
    timestamp,
    ROUND(AVG(value),2) AS avg_current
FROM smartmeter.gold.power_quality
WHERE parameter_name = 'Current (L1)'
GROUP BY timestamp
ORDER BY timestamp;
```

```python
%sql
SELECT
    timestamp,
    AVG(value) AS avg_active_power
FROM smartmeter.gold.power_quality
WHERE parameter_name = 'Active power'
GROUP BY timestamp
ORDER BY timestamp;
```

```python
%sql
SELECT
    area_name,
    COUNT(*) AS reading_count
FROM smartmeter.gold.power_quality
GROUP BY area_name
ORDER BY reading_count DESC;
```

```python
%sql
SELECT
    quality,
    COUNT(*) AS reading_count
FROM smartmeter.gold.power_quality
GROUP BY quality
ORDER BY reading_count DESC;
```
