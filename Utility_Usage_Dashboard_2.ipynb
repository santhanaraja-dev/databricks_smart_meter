```python
%sql
SELECT
    COUNT(*) AS total_readings
FROM smartmeter.gold.utility_usage;
```

```python
%sql
SELECT
    COUNT(DISTINCT meter_id) AS reporting_meters
FROM smartmeter.gold.utility_usage;
```

```python
%sql
SELECT
    meter_type,
    COUNT(*) AS reading_count
FROM smartmeter.gold.utility_usage
GROUP BY meter_type
ORDER BY reading_count DESC;
```

```python
%sql
SELECT
    DATE(timestamp) AS reading_date,
    COUNT(*) AS reading_count
FROM smartmeter.gold.utility_usage
GROUP BY DATE(timestamp)
ORDER BY reading_date;
```

```python
%sql
SELECT
    area_name,
    COUNT(*) AS reading_count
FROM smartmeter.gold.utility_usage
GROUP BY area_name
ORDER BY reading_count DESC;
```

```python
%sql
SELECT
    parameter_name,
    COUNT(*) AS reading_count
FROM smartmeter.gold.utility_usage
GROUP BY parameter_name
ORDER BY reading_count DESC;
```
