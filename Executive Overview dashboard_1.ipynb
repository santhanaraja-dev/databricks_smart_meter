```python
%sql
SELECT COUNT(*) AS total_meters
FROM smartmeter.gold.meter_overview;
```

```python
%sql
SELECT COUNT(*) AS active_meters
FROM smartmeter.gold.meter_overview
WHERE status = 'ACTIVE';
```

```python
%sql
SELECT COUNT(*) AS faulty_meters
FROM smartmeter.gold.meter_overview
WHERE status = 'FAULT';
```

```python
%sql
SELECT COUNT(*) AS total_alarms
FROM smartmeter.gold.alarm_summary;
```

```python
%sql
SELECT
    ROUND(
        SUM(good_readings) * 100.0 / SUM(total_readings),
        2
    ) AS data_quality_pct
FROM smartmeter.gold.data_quality;
```

```python
%sql
SELECT
    meter_type,
    COUNT(*) AS meter_count
FROM smartmeter.gold.meter_overview
GROUP BY meter_type
ORDER BY meter_count DESC;
```

```python
%sql
SELECT
    area_name,
    COUNT(*) AS meter_count
FROM smartmeter.gold.meter_overview
GROUP BY area_name
ORDER BY meter_count DESC;
```

```python
%sql
SELECT
    severity,
    SUM(alarm_count) AS alarm_count
FROM smartmeter.gold.alarm_summary
GROUP BY severity
ORDER BY alarm_count DESC;
```

```python
%sql
SELECT
    status,
    COUNT(*) AS meter_count
FROM smartmeter.gold.meter_overview
GROUP BY status
ORDER BY meter_count DESC;
```
