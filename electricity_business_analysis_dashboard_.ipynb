```python
%sql
SELECT
    COUNT(DISTINCT meter_id) AS total_electricity_meters
FROM smartmeter.gold.electricity_power_analysis;
```

```python
%sql
SELECT
    ROUND(AVG(active_power_kw), 2) AS avg_active_power_kw
FROM smartmeter.gold.electricity_power_analysis;
```

```python
%sql
SELECT
    ROUND(MAX(active_power_kw), 2) AS peak_active_power_kw
FROM smartmeter.gold.electricity_power_analysis;
```

```python
%sql
SELECT
    hour,
    ROUND(AVG(avg_power_kw), 2) AS average_power_kw,
    ROUND(MAX(peak_power_kw), 2) AS peak_power_kw
FROM smartmeter.gold.hourly_electricity_power
GROUP BY hour
ORDER BY hour;
```

```python
%sql
SELECT
    area_name,
    ROUND(AVG(avg_power_kw), 2) AS average_power_kw,
    ROUND(MAX(peak_power_kw), 2) AS peak_power_kw
FROM smartmeter.gold.electricity_area_analysis
GROUP BY area_name
ORDER BY average_power_kw DESC;
```

```python
%sql
SELECT
    quality,
    COUNT(*) AS reading_count
FROM smartmeter.gold.electricity_power_analysis
GROUP BY quality
ORDER BY reading_count DESC;
```
