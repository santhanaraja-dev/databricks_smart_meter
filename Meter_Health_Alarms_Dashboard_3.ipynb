```python
%sql
SELECT
    COUNT(*) AS faulty_meters
FROM smartmeter.gold.meter_health
WHERE status = 'FAULT';
```

```python
%sql
SELECT
    COUNT(DISTINCT meter_id) AS meters_with_alarms
FROM smartmeter.gold.meter_health
WHERE alarm_count > 0;
```

```python
%sql
SELECT
    SUM(unacknowledged_count) AS unacknowledged_alarms
FROM smartmeter.gold.alarm_summary;
```

```python
%sql
SELECT
    severity,
    SUM(alarm_count) AS alarm_count
FROM smartmeter.gold.alarm_summary
GROUP BY severity
ORDER BY alarm_count DESC;
```

```python
%sql
SELECT
    area_name,
    SUM(alarm_count) AS alarm_count
FROM smartmeter.gold.alarm_summary
GROUP BY area_name
ORDER BY alarm_count DESC;
```

```python
%sql
SELECT
    'Acknowledged' AS alarm_status,
    SUM(acknowledged_count) AS alarm_count
FROM smartmeter.gold.alarm_summary

UNION ALL

SELECT
    'Unacknowledged' AS alarm_status,
    SUM(unacknowledged_count) AS alarm_count
FROM smartmeter.gold.alarm_summary;
```

```python
%sql
SELECT
    alarm_type,
    SUM(alarm_count) AS alarm_count
FROM smartmeter.gold.alarm_summary
GROUP BY alarm_type
ORDER BY alarm_count DESC;
```
